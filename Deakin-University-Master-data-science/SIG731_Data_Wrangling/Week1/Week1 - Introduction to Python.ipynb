## Calling libraries

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import warnings
warnings.filterwarnings('ignore')
%matplotlib inline
```

## Reading in data 

**Here's a brief explanation of the dataset used**

* For the purposes of this session, we are using a simple dataset in order to learn more about visualization and plotting data.
* The dataset consists of one numeric column, called sale_price.
* The sale_price refers to the property value in dollars.

```python
df = np.loadtxt('sale_price.csv', dtype = 'float', delimiter = ',')
```

## Histogram plot
**When to use** :- A histogram is a great tool for quickly assessing a `probability distribution` that is intuitively understood by almost any audience. Python offers a handful of different options for building and plotting histograms. Most people know a histogram by its graphical representation, which is similar to a bar graph:

**Y-axis - Counts**

**X-axis - Continous Variable**

```python
plt.hist(df, bins = 10, color = 'purple')
plt.title("Plotting a histogram using matplotlib")
plt.show()
```

```python
sns.histplot(df, bins = 10, color = 'red')
plt.title("Plotting a histogram using seaborn")
plt.show()
```

## Distplot (Distribution plot) .

**what is an distplot.**

The most convenient way to take a quick look at a distribution in seaborn is the distplot() function. By default, this will draw a histogram and fit a kernel density estimate (KDE).

```python
sns.distplot(df,bins=10,kde=True);   
#this is an histogram representation of distribution of values and bin size sets the size of individual bin.

plt.title("Plotting a distribution plot using seaborn")
plt.show()
```

We can intrepret from the above plot that in the saleprice distribution where the mostly purchased house was around 15 lakhs.

**Rugplot/distplot**:-Histograms are likely familiar, and a hist function already exists in matplotlib. A histogram represents the distribution of data by forming bins along the range of the data and then drawing bars to show the number of observations that fall in each bin. 

* Let’s remove the density curve and add a rug plot, which draws a small vertical tick at each observation. You can make the rug plot itself with the rugplot() function, but it is also available in distplot()

```python
sns.distplot(df,bins=50,kde=False,rug=True)
plt.title("Plotting a rugplot alongside a distribution plot using seaborn")
plt.show()
```

From the above plot we can interpret that most of the houses were purchased by the customers with price ranging between 1 lakh and 3 lakh
