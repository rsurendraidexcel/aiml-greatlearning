## Donloading and Extrating Data 

```python
## Data Extration Download method and return the Pandas dataframe
def nhanes_data_extraction(url):
    response = requests.get(url)
    # Save the file locally
    with open("temp_data.xpt", "wb") as f:
        f.write(response.content)
    # Read the .XPT file using pyreadstat
    df, meta = pyreadstat.read_xport("temp_data.xpt" ,encoding='latin1')
    # Display the first few rows
    return df
```

- Define nhanes_data_extraction() method  this method will Download data specify url
- write to local disk "temp_data.xpt" file 
- Read the .XPT file using pyreadstat os  pyreadstat pakage has to install !pip install pyreadstat
- Return Panda dataframe download file

### Extracting data 2021-2023 National Helath and Nutrition Examination Center (Nhames ) in padas dataframe

```python
# Demographics
df_demo_2021_2023 = nhanes_data_extraction(url_demo_2021_2023)

# Body Measures
df_bmx_2021_2023 = nhanes_data_extraction(url_bmx_2021_2023)

# Diatary data review
df_dr1review_2021_2023 = nhanes_data_extraction(url_dr1review_2021_2023)

#Laboratory Data: alpha-1-Acid Glycoprotein
df_acid_glycoprotin_2021_2023 = nhanes_data_extraction(url_acid_glycoprotin_2021_2023)

# Questionnaire Data: Alcohol Use data load
df_alc_2021_2023 = nhanes_data_extraction(url_alc_2021_2023)
```

## Data Cleaning and Preparation

```python
# Merge demographics with body measures for 2021-2023
df_2021_2023 = pd.merge(df_demo_2021_2023, df_bmx_2021_2023, on='SEQN')
df_2021_2023 = pd.merge(df_2021_2023, df_alc_2021_2023, on='SEQN')

# Merge dietary, AGP, and demographics data
df_2021_2023 = pd.merge(df_2021_2023, df_dr1review_2021_2023, on='SEQN')
df_2021_2023 = pd.merge(df_2021_2023, df_acid_glycoprotin_2021_2023, on='SEQN')


# Merge demographics with body measures for 2017-2020
df_2017_2020 = pd.merge(df_demo_2017_2020, df_bmx_2017_2020, on='SEQN')
df_2017_2020 = pd.merge(df_2017_2020, df_alc_2017_2020, on='SEQN')

# Merge dietary, AGP, and demographics data
df_2017_2020 = pd.merge(df_2017_2020, df_dr1review_2021_2023, on='SEQN')
df_2017_2020 = pd.merge(df_2017_2020, df_acid_glycoprotin_2021_2023, on='SEQN')


# Combine both datasets for comparison 2021-2023 and 2017-2020
df_combined = pd.concat([df_2021_2023, df_2017_2020], ignore_index=True)

# Select relevant columns bmx_acl
df_combined_demo_bmx_acl = df_combined[['RIAGENDR', 'RIDAGEYR', 'BMXBMI', 'ALQ121', 'ALQ151']]

# Rename columns for clarity
df_combined_demo_bmx_acl.rename(columns={
    'RIAGENDR': 'Gender',
    'RIDAGEYR': 'Age',
    'BMXBMI': 'BMI',
    'ALQ121': 'Alcohol_Consumption',
    'ALQ151': 'Alcohol_Frequency'
}, inplace=True)

# Handle missing values
df_combined_demo_bmx_acl.dropna(inplace=True)
df_combined_demo_bmx_acl
```

- dataFrame df_2021_2023 Merge demographics with body measures for 2021-2023
- dataFrame df_2021_2023 Merge dietary, AGP, and demographics data
- dataFrame df_2017_2020 Merge demographics with body measures for 2017-2020
- dataFrame df_2017_2020 Merge dietary, AGP, and demographics data
- df_combined dataFraem Combine both datasets for comparison 2021-2023 and 2017-2020
- Select relevant columns bmx_acl
- Rename columns for clarity
- Handle missing values

```python
# Select relevant columns diary and apg
df_diet_agp = df_combined[['RIAGENDR', 'RIDAGEYR', 'DR1IKCAL', 'DR1IPROT', 'DR1ICARB', 'DR1ISUGR', 'DR1ITFAT', 'LBXAGP']]

# Rename columns for clarity
df_diet_agp.rename(columns={
    'RIAGENDR': 'Gender',
    'RIDAGEYR': 'Age',
    'DR1IKCAL': 'Calories',
    'DR1IPROT': 'Protein',
    'DR1ICARB': 'Carbohydrates',
    'DR1ISUGR': 'Sugars',
    'DR1ITFAT': 'Total_Fat',
    'LBXAGP': 'AGP'
}, inplace=True)

#Handle missing values
df_diet_agp.dropna(inplace=True)
df_diet_agp
```

- Handle missing values

## Create Visualizations

### Distribution of BMI by Gender

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Gender', y='BMI', data=df_combined_demo_bmx_acl)
plt.title('Distribution of BMI by Gender')
plt.xlabel('Gender (1: Male, 2: Female)')
plt.ylabel('BMI')
plt.show()
```

**Insight:**

This plot shows the distribution of BMI across genders. It can help identify if there are significant differences in BMI between males and females.

### Alcohol Consumption by Age Group

```python
df_combined_demo_bmx_acl['Age_Group'] = pd.cut(df_combined_demo_bmx_acl['Age'], bins=[0, 20, 40, 60, 80, 100], labels=['0-20', '21-40', '41-60', '61-80', '81-100'])
plt.figure(figsize=(10, 6))
sns.countplot(x='Age_Group', hue='Alcohol_Consumption', data=df_combined_demo_bmx_acl)
plt.title('Alcohol Consumption by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.legend(title='Alcohol Consumption (1: Yes, 2: No)')
plt.show()
```

**Insight:**

This plot shows the prevalence of alcohol consumption across different age groups. It can help identify trends in alcohol use by age.

 ## Correlation Heatmap of BMI and alcohol consumption

```python
plt.figure(figsize=(10, 6))
bmi_alc = df_combined_demo_bmx_acl[['Gender','BMI','Age','Alcohol_Consumption','Alcohol_Frequency']]
sns.heatmap(bmi_alc.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

**Insight:**

This heatmap shows the correlation between different variables (BMI, Age, Alcohol Consumption). It can help identify relationships between variables. AG-Group is Categorical data we avoid here

### BMI Distribution by Alcohol Consumption

```python
plt.figure(figsize=(10, 6))
sns.violinplot(x='Alcohol_Consumption', y='BMI', data=df_combined_demo_bmx_acl)
plt.title('BMI Distribution by Alcohol Consumption')
plt.xlabel('Alcohol Consumption (1: Yes, 2: No)')
plt.ylabel('BMI')
plt.show()
```

**Insight:**

This plot shows the distribution of BMI for individuals who consume alcohol versus those who do not. It can help identify if alcohol consumption is associated with differences in BMI.

### Alcohol Frequency by Gender

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='Gender', hue='Alcohol_Frequency', data=df_combined_demo_bmx_acl)
plt.title('Alcohol Frequency by Gender')
plt.xlabel('Gender (1: Male, 2: Female)')
plt.ylabel('Count')
plt.legend(title='Alcohol Frequency')
plt.show()
```

**Insight:**

This plot shows the frequency of alcohol consumption by gender. It can help identify if there are differences in drinking patterns between males and females.

###  Distribution of Alpha-1-Acid Glycoprotein (AGP)

```python
plt.figure(figsize=(10, 6))
sns.histplot(df_diet_agp['AGP'], bins=30, kde=True, color='blue')
plt.title('Distribution of Alpha-1-Acid Glycoprotein (AGP)')
plt.xlabel('AGP (mg/dL)')
plt.ylabel('Frequency')
plt.show()
```

**Insight:**

This plot shows the distribution of AGP levels in the population. AGP is a marker of inflammation, and higher levels may indicate chronic inflammation or immune response. The distribution can help identify if there are outliers or a skewed distribution.

### Relationship Between Caloric Intake and AGP

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Calories', y='AGP', data=df_diet_agp, alpha=0.6, color='green')
plt.title('Relationship Between Caloric Intake and AGP')
plt.xlabel('Daily Caloric Intake (kcal)')
plt.ylabel('AGP (mg/dL)')
plt.show()
```

**Insight:**

This scatterplot explores whether there is a relationship between daily caloric intake and AGP levels. Higher caloric intake might be associated with higher inflammation markers, but this would require further statistical analysis to confirm.

### Macronutrient Composition by AGP Levels

We can group individuals by AGP levels (low, medium, high) and compare their macronutrient intake (protein, carbohydrates, fats).


```python
# Create AGP categories
df_diet_agp['AGP_Category'] = pd.cut(df_diet_agp['AGP'], bins=[0, 50, 100, 150, 200], labels=['Low', 'Medium', 'High', 'Very High'])

# Plot macronutrient composition by AGP category
plt.figure(figsize=(12, 6))
sns.boxplot(x='AGP_Category', y='Protein', data=df_diet_agp, color='blue', label='Protein')
sns.boxplot(x='AGP_Category', y='Carbohydrates', data=df_diet_agp, color='orange', label='Carbohydrates')
sns.boxplot(x='AGP_Category', y='Total_Fat', data=df_diet_agp, color='green', label='Total Fat')
plt.title('Macronutrient Composition by AGP Levels')
plt.xlabel('AGP Category')
plt.ylabel('Intake (grams)')
plt.legend()
plt.show()
```

**Insight:**

This plot compares the intake of protein, carbohydrates, and fats across different AGP levels. It can help identify if certain macronutrient patterns are associated with higher or lower inflammation markers.

### Sugar Intake vs. AGP Levels

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Sugars', y='AGP', data=df_diet_agp, alpha=0.6, color='red')
plt.title('Sugar Intake vs. AGP Levels')
plt.xlabel('Daily Sugar Intake (grams)')
plt.ylabel('AGP (mg/dL)')
plt.show()
```

**Insight:**

This scatterplot explores the relationship between sugar intake and AGP levels. High sugar intake is often linked to inflammation, so this visualization can help identify if there is a trend.


### AGP Levels by Age and Gender

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x='Age', y='AGP', hue='Gender', data=df_diet_agp)
plt.title('AGP Levels by Age and Gender')
plt.xlabel('Age (years)')
plt.ylabel('AGP (mg/dL)')
plt.legend(title='Gender (1: Male, 2: Female)')
plt.show()
```

**Insight:**

This plot shows how AGP levels vary by age and gender. It can help identify if certain age groups or genders are more prone to higher inflammation markers.
