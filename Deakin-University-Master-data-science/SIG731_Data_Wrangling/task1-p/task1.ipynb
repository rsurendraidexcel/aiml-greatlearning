## Q1. Generate 10 individuals’ weekly salaries using the given distribution.

````python
def generate_weekly_salaries(mean, std_dev, num_individuals):
    salaries = []
    for _ in range(num_individuals):
        # Use random.gauss to generate salaries based on mean and std deviation
        salary = random.gauss(mean, std_dev)
        salaries.append(round(salary, 2))  # Round to 2 decimal places
    return salaries

# Given Parameters
mean_salary = 1431
std_dev_salary = 527
num_individuals = 10

# weekly salaries
weekly_salaries = generate_weekly_salaries(mean_salary, std_dev_salary, num_individuals)

# List of salaries
print("Generated Weekly Salaries:", weekly_salaries)
````

## Q2: Calculate annual income tax for each individual based on their weekly salary

````python
def calculate_annual_income_tax(weekly_salaries):
    income_taxes = []
    for weekly_salary in weekly_salaries:
        annual_income = weekly_salary * 52
        if annual_income <= 18200:
            tax = 0
        elif annual_income <= 45000:
            tax = 0.19 * (annual_income - 18200)
        elif annual_income <= 120000:
            tax = 5092 + 0.325 * (annual_income - 45000)
        elif annual_income <= 180000:
            tax = 29467 + 0.37 * (annual_income - 120000)
        else:
            tax = 51667 + 0.45 * (annual_income - 180000)
        income_taxes.append(round(tax, 2))
    return income_taxes

# Calculate income tax for each individual
income_taxes = calculate_annual_income_tax(weekly_salaries)

# List of Income Task
print("Annual Income Taxes:", income_taxes)
````

### Q3: Calculate weekly withholding tax for each individual based on their weekly salary

````python
def calculate_weekly_withholding_tax(weekly_salaries):
    withholding_taxes = []
    for weekly_salary in weekly_salaries:
        x = weekly_salary + 0.99  # Weekly salary plus 99 cents
        if x < 359:
            tax = 0
        elif x < 438:
            tax = 0.19 * x - 68.3462
        elif x < 548:
            tax = 0.29 * x - 112.1942
        elif x < 721:
            tax = 0.21 * x - 68.3465
        elif x < 865:
            tax = 0.219 * x - 74.8369
        elif x < 1282:
            tax = 0.3477 * x - 186.2119
        elif x < 2307:
            tax = 0.345 * x - 182.7504
        elif x < 3461:
            tax = 0.39 * x - 286.5965
        else:
            tax = 0.47 * x - 563.5196
        withholding_taxes.append(round(tax, 2))
    return withholding_taxes

# Calculate withholding tax for each individual
withholding_taxes = calculate_weekly_withholding_tax(weekly_salaries)

# Output the withholding tax values
print("Weekly Withholding Taxes:", withholding_taxes)
````

## Q4: Calculate the annual eligible tax refund for each individual

````python
def calculate_tax_return(income_taxes, withholding_taxes):
    tax_returns = []
    for income_tax, weekly_withholding_tax in zip(income_taxes, withholding_taxes):
        annual_withholding_tax = weekly_withholding_tax * 52
        tax_return = annual_withholding_tax - income_tax
        tax_returns.append(round(tax_return, 2))
    return tax_returns

# Calculate tax returns for each individual
tax_returns = calculate_tax_return(income_taxes, withholding_taxes)


# Output List of Annual Tax for each Indivisual
print("Annual Tax Returns:", tax_returns)
print("--------------------------"*4)


# Print detailed information for each individual
for i, (weekly_salary, weekly_withholding_tax, income_tax, tax_return) in enumerate(zip(weekly_salaries, withholding_taxes, income_taxes, tax_returns), start=1):
    weekly_income = weekly_salary - weekly_withholding_tax
    print(f"## Person {i}")
    print(f"Weekly salary: ${weekly_salary:.2f}")
    print(f"Weekly withholding tax: ${weekly_withholding_tax:.2f}")
    print(f"Weekly income: ${weekly_income:.2f}")
    print(f"Income tax: ${income_tax:.2f}")
    print(f"Tax return: ${tax_return:.2f}")
    print("----------------"*3)
````

### Q5. Calculate the employer’s weekly before-tax superannuation contribution.

````python
#Calculate weekly before-tax superannuation contributions
def calculate_super_contributions(weekly_salaries, contribution_rate=0.11):
    super_contributions = []
    for salary in weekly_salaries:
        contribution = salary * contribution_rate
        super_contributions.append(contribution)
    return super_contributions

# Calculate and print superannuation contributions
super_contributions = calculate_super_contributions(weekly_salaries)
for i, contribution in enumerate(super_contributions, start=1):
    print(f"## Person {i} weekly superannuation contribution\n${contribution:.2f}")
````

### Q6. Repeat Subtasks 2, 3, 4 to recalculate the income tax, withholding tax, and tax return.

````python
#Calculate the Yearly Income tax 
def calculate_yearly_income_tax(taxable_income):
    if taxable_income <= 18200:
        return 0
    elif taxable_income <= 45000:
        return 0.19 * (taxable_income - 18200)
    elif taxable_income <= 120000:
        return 5092 + 0.325 * (taxable_income - 45000)
    elif taxable_income <= 180000:
        return 29467 + 0.37 * (taxable_income - 120000)
    else:
        return 51667 + 0.45 * (taxable_income - 180000)
        
#Calculating taxholding base on weekly salary
def calculate_withholding_tax(weekly_salary):
    if weekly_salary < 359:
        return 0
    elif weekly_salary < 438:
        return 0.19 * (weekly_salary + 0.99) - 68.3462
    elif weekly_salary < 548:
        return 0.29 * (weekly_salary + 0.99) - 112.1942
    elif weekly_salary < 721:
        return 0.21 * (weekly_salary + 0.99) - 68.3465
    elif weekly_salary < 865:
        return 0.219 * (weekly_salary + 0.99) - 74.8369
    elif weekly_salary < 1282:
        return 0.3477 * (weekly_salary + 0.99) - 186.2119
    elif weekly_salary < 2307:
        return 0.345 * (weekly_salary + 0.99) - 182.7504
    elif weekly_salary < 3461:
        return 0.39 * (weekly_salary + 0.99) - 286.5965
    else:
        return 0.47 * (weekly_salary + 0.99) - 563.5196


def calculate_tax_return(annual_income_tax, annual_withholding_tax):
    return annual_withholding_tax - annual_income_tax


def calculate_superannuation(weekly_salary, rate=0.11):
    return weekly_salary * rate    

# List of superannuation contributions
super_contributions = [calculate_superannuation(salary) for salary in weekly_salaries]

# Recalculate income tax, withholding tax, and tax return considering superannuation as a deduction
results = []
for i, weekly_salary in enumerate(weekly_salaries):
    super_contribution = super_contributions[i]
    taxable_income = (weekly_salary - super_contribution) * 52
    annual_income_tax = calculate_yearly_income_tax(taxable_income)
    weekly_withholding_tax = calculate_withholding_tax(weekly_salary - super_contribution)
    annual_withholding_tax = weekly_withholding_tax * 52
    tax_return = calculate_tax_return(annual_income_tax, annual_withholding_tax)

    results.append({
        "Weekly Salary": weekly_salary,
        "Weekly Superannuation": super_contribution,
        "Weekly Taxable Income": weekly_salary - super_contribution,
        "Annual Income Tax": annual_income_tax,
        "Weekly Withholding Tax": weekly_withholding_tax,
        "Tax Return": tax_return,
        "Weekly Income After Tax": weekly_salary - weekly_withholding_tax - super_contribution
    })

#Output of Result
for i, result in enumerate(results, start=1):
    print(f"## Person {i}")
    print(f"Weekly salary (including superannuation): ${result['Weekly Salary']:.2f}")
    print(f"Weekly superannuation contribution: ${result['Weekly Superannuation']:.2f}")
    print(f"Weekly taxable income: ${result['Weekly Taxable Income']:.2f}")
    print(f"Weekly withholding tax: ${result['Weekly Withholding Tax']:.2f}")
    print(f"Weekly income after tax: ${result['Weekly Income After Tax']:.2f}")
    print(f"Annual income tax: ${result['Annual Income Tax']:.2f}")
    print(f"Tax return: ${result['Tax Return']:.2f}")
    print("-------------------------"*3)
````
