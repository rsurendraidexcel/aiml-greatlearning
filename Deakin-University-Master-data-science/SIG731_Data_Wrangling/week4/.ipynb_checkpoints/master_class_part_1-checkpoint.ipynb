<span style="font-family: Arial; font-weight:bold;font-size:1.50em;color:#1D1B9E;"> Series Ex:1 Combine DataFrame objects

```python
Table1 = pd.DataFrame({'A': ['10', '11', '12', '13'],
                    'B': ['20', '21', '22', '23'],
                    'C': ['30', '31', '32', '33'],
                    'D': ['40', '41', '42', '43']},index=[0, 1, 2, 3])
Table1
```

```python
Table2 = pd.DataFrame({'A': ['14', '15', '16', '17'],
                    'B': ['24', '25', '26', '27'],
                    'C': ['34', '35', '36', '37'],
                    'D': ['44', '45', '46', '47']},index=[4, 5, 6, 7])
Table2
```

```python
Table3 = pd.DataFrame({'A': ['18', '19'],
                    'B': ['28', '29'],
                    'C': ['38', '39'],
                    'D': ['48', '49']},index=[8, 9])
Table3
```

```python
Table4 = pd.DataFrame({'A': ['18', '19'],
                    'B': ['28', '29'],
                    'Y': ['38', '39'],
                    'Z': ['48', '49']},index=[8, 9])
Table4
```

```python
df = [Table1, Table2, Table3]
df
```

```python
pd.concat([Table1, Table2, Table3], axis=0, join='outer', ignore_index=True,
          keys=None, levels=None, names=None, verify_integrity=False,
          copy=True)
```

```python
Data = pd.concat([Table1, Table2, Table3])
Data
```

<span style="font-family: Arial; font-weight:bold;font-size:1.50em;color:#1D1B9E;"> Concat Ex:2 Adding hierarchical index

```python
pd.concat(df,keys=['table1', 'table2', 'table3'] )
```

<span style="font-family: Arial; font-weight:bold;font-size:1.50em;color:#1D1B9E;"> Series Ex:3 Combine DataFrames with overlapping columns and return only those that are shared

```python
Output = pd.concat([Table1, Table2], axis=1, join='outer')#,join_axes=[Table1.index])
Output
```

```python
Output = pd.concat([Table1, Table4], axis=0)
Output
```

```python
Output = pd.concat([Table2, Table1], axis=1)
Output
```

<span style="font-family: Arial; font-weight:bold;font-size:2.00em;color:#9F1F1B;"> II. Merge

```python
Image('2.png')
```

```python
company_1 = pd.DataFrame({'emp_id': range(5),
                           'emp_name': ['a', 'b', 'c', 'd', 'e']})
company_1
```

```python
company_2 = pd.DataFrame({'emp_id': range(2, 7),
                           'emp_name': ['f', 'g', 'h', 'i', 'j']})
company_2
```

```python
company_3 = pd.DataFrame({'employee_id': range(4,9),
                           'emp_name': ['f', 'g', 'h', 'i', 'j']})
company_3
```

<span style="font-family: Arial; font-weight:bold;font-size:2.00em;color:#9F1F1B;">  Inner Join

```python
Image('2.1.png')
```

\

```python
pd.merge(company_1, company_3, left_on='emp_id', right_on = 'employee_id',how='inner')
```

<span style="font-family: Arial; font-weight:bold;font-size:2.00em;color:#9F1F1B;"> Outer Join

```python
Image('2.2.png')
```

```python
pd.merge(company_1, company_2, on='emp_id', how='outer')
```

<span style="font-family: Arial; font-weight:bold;font-size:2.00em;color:#9F1F1B;"> Right Join

```python
Image('2.3.png')
```

```python
pd.merge(company_1, company_2, on='emp_id', how='right')
```

<span style="font-family: Arial; font-weight:bold;font-size:2.00em;color:#9F1F1B;"> Left Join

```python
Image('2.4.png')
```

```python
pd.merge(company_1, company_2, on='emp_id', how='left')
```

<span style="font-family: Arial; font-weight:bold;font-size:2.00em;color:#9F1F1B;"> VI. Melt

```python
Image('3.3.png')
```

```python
data = {
    "location":['Bangalore', 'Chennai'],
    "Temperature" : ["Predict", "Actual"],
    "Jan-2019":[30, 32],
    "Feb-2019":[45, 43],
    "Mar-2019":[24, 22]
}
df = pd.DataFrame(data, columns=['location', 'Temperature','Jan-2019', 'Feb-2019', 'Mar-2019'])
print(df)
```

```python
df2 = pd.melt(df, id_vars=["location","Temperature"], var_name="Month_year", value_name="Temp_value")
print(df2)
```

<span style="font-family: Arial; font-weight:bold;font-size:2.00em;color:#9F1F1B;"> VII. Pivot Table

```python
#Create a DataFrame
d = {
    'Name':['Jony','Anil','Sunil','Jony','Anil','Rocky',
            'Jony','Anil','Sunil','Jony','Anil','Rocky'],
    'Exam_Months':['3_Months','3_Months','3_Months','3_Months','3_Months','3_Months',
            '6_Months','6_Months','6_Months','6_Months','6_Months','6_Months'],

    'Subject':['Python','Python','SQL','SQL','ML','ML',
               'Python','Python','SQL','SQL','ML','ML'],
   'Score':[62,47,55,74,31,77,85,63,42,67,89,81]}

df = pd.DataFrame(d,columns=['Name','Exam_Months','Subject','Score'])
df


```

```python
df.groupby(['Name','Subject'])['Score'].mean()
```

```python
Image('4.png')
```

<span style="font-family: Arial; font-weight:bold;font-size:1.50em;color:#1D1B9E;"> Pivot Ex:1 Create pivot table with aggregate function mean

```python
pd.pivot_table(df,values = 'Score', index=['Name','Subject'], aggfunc = 'mean')
```

<span style="font-family: Arial; font-weight:bold;font-size:1.50em;color:#1D1B9E;"> Pivot Ex:2 Create pivot table with aggregate function sum

```python
df.groupby(['Subject','Exam_Months'])['Score'].sum()
```

```python
pd.pivot_table(df, index=['Name','Subject'], aggfunc='sum', margins=True, margins_name = "Total")
```
